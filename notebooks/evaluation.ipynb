# Temporal Reasoning Engine — Evaluation Notebook

This notebook demonstrates and reproduces the evaluation of the symbolic **Temporal Reasoning Engine** over the 150-example benchmark dataset.

```python
import sys
from pathlib import Path

# Ensure src is in python path
src_path = str(Path("../src").resolve())
if src_path not in sys.path:
    sys.path.insert(0, src_path)

from temporal_reasoning import (
    TemporalReasoningEngine,
    BenchmarkEvaluator,
    TemporalRelation,
    __version__
)

print(f"Loaded Temporal Reasoning Engine v{__version__}")
```

## 1. Quick Demonstration of Symbolic Reasoning

```python
engine = TemporalReasoningEngine()
engine.add_statement("Alice arrived before Bob.")
engine.add_statement("Bob left before Charlie.")

rel = engine.query("Alice", "Charlie")
prov = engine.query_explanation("Alice", "Charlie")

print(f"Query (Alice -> Charlie): {rel.value.upper()}")
print(f"Reasoning Provenance:     {prov}")
print(f"Graph Consistent:         {engine.is_consistent()}")
```

## 2. Benchmark Evaluation

Run the full 150-example benchmark across all 8 temporal reasoning categories:

```python
evaluator = BenchmarkEvaluator()
benchmark_path = Path("../data/benchmark.json").resolve()
metrics = evaluator.evaluate_file(benchmark_path)

print(metrics.summary_report())
```

## 3. Metrics Summary Table

```python
import pandas as pd

cat_data = [
    {"Category": cat, "Total": stats["total"], "Correct": stats["correct"], "Accuracy (%)": stats["accuracy"] * 100}
    for cat, stats in sorted(metrics.category_breakdown.items())
]
df_cats = pd.DataFrame(cat_data)
df_cats
```
